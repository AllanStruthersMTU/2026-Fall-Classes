# Rössler sensitivity from two separate ODE solves

This notebook deliberately separates the calculation into two ODE solves. First we solve the Rössler system and obtain a dense numerical interpolant $\mathbf{x}(t)$. We then treat that interpolant as a known, time-dependent input to the variational equation for the initial-condition sensitivity matrix

$$J(t)=\frac{\partial \mathbf{x}(t)}{\partial \mathbf{x}_0}, \qquad \dot J(t)=Df(\mathbf{x}(t))J(t), \qquad J(0)=I.$$

This is pedagogically useful because the state trajectory and tangent dynamics remain visibly distinct.

## 1. Packages

If needed, uncomment the installation line once. `DifferentialEquations` supplies `ODEProblem`, `solve`, and `Tsit5`; `LinearAlgebra` supplies the identity matrix and singular-value decomposition; and `Plots` supplies the figures.

```julia
# using Pkg; Pkg.add(["DifferentialEquations", "Plots"])
using DifferentialEquations
using LinearAlgebra
using Plots
```

## 2. First solve: the Rössler trajectory

We use the standard Rössler equations

$$\dot x=-y-z, \qquad \dot y=x+ay, \qquad \dot z=b+z(x-c),$$

with $(a,b,c)=(0.2,0.2,5.7)$. The solution object returned by DifferentialEquations.jl is callable. Thus, after solving, `state_sol(t)` evaluates the solver's continuous interpolant at any time inside `tspan`. The second solve will use exactly this feature.

```julia
function rossler!(du, u, p, t)
    a, b, c = p
    x, y, z = u

    du[1] = -y - z
    du[2] =  x + a*y
    du[3] =  b + z*(x - c)
    return nothing
end

rossler_parameters = (0.2, 0.2, 5.7)
state_initial = [1.0, 0.0, 0.0]
tspan = (0.0, 50.0)

state_problem = ODEProblem(rossler!, state_initial, tspan, rossler_parameters)

state_sol = solve(
    state_problem,
    Tsit5();
    abstol = 1e-10,
    reltol = 1e-10,
    dense = true,
    saveat = 0.02
)
```

The `saveat` values are convenient for plotting, but they are not the only times at which the solution can be evaluated. With dense output enabled, `state_sol(1.234)` evaluates an interpolating polynomial constructed by the solver.

```julia
t_test = 1.234
println("Interpolated state at t = $t_test:")
display(state_sol(t_test))

trajectory_plot = plot(
    state_sol[1, :], state_sol[2, :], state_sol[3, :];
    xlabel = "x", ylabel = "y", zlabel = "z",
    title = "Rössler trajectory: first solve",
    label = false, linewidth = 1.2, color = :navy,
    camera = (35, 25), size = (750, 500)
)
```

## 3. Derive the variational equation

For $\dot{\mathbf{x}}=f(\mathbf{x})$, the matrix governing an infinitesimal perturbation is

$$Df(x,y,z)=\begin{pmatrix}0&-1&-1\\1&a&0\\z&0&x-c\end{pmatrix}.$$

The sensitivity matrix is not the same object as $Df$. The matrix $Df(\mathbf{x}(t))$ gives the instantaneous linearization at the current state. The matrix $J(t)$ accumulates that linearized stretching from the initial time to time $t$. It obeys

$$\dot J(t)=Df(\mathbf{x}(t))J(t).$$

Because $J$ measures sensitivity to the initial condition, $J(0)=I_3$. In the second solve, `state_interpolant(t)` supplies the values of $x(t)$ and $z(t)$ needed in $Df$.

```julia
function sensitivity_jacobian!(dJ_vector, J_vector, p, t)
    state_interpolant, rossler_parameters = p
    a, b, c = rossler_parameters

    # This evaluates the dense interpolant produced by the FIRST solve.
    x, y, z = state_interpolant(t)

    # The second ODE is represented as nine scalar equations, while these
    # reshaped views let us write and interpret them as a 3 × 3 matrix.
    J  = reshape(J_vector, 3, 3)
    dJ = reshape(dJ_vector, 3, 3)

    # Compute dJ = Df(x(t))*J explicitly, one column at a time.
    # This also avoids allocating Df at every right-hand-side evaluation.
    for column in 1:3
        dJ[1, column] = -J[2, column] - J[3, column]
        dJ[2, column] =  J[1, column] + a*J[2, column]
        dJ[3, column] =  z*J[1, column] + (x - c)*J[3, column]
    end

    return nothing
end
```

## 4. Second solve: the sensitivity matrix

The parameter passed to this ODE is the tuple `(state_sol, rossler_parameters)`. Therefore the second solver can request the Rössler state at each of its own internal time steps. Its steps do not have to coincide with the saved points from the first solve.

The tolerances for the first solve are chosen tightly because interpolation error from the first solve becomes input error in the second. Both solves use `Tsit5()` as requested.

```julia
J_initial = Matrix{Float64}(I, 3, 3)
J_initial_vector = vec(J_initial)

sensitivity_parameters = (state_sol, rossler_parameters)

sensitivity_problem = ODEProblem(
    sensitivity_jacobian!,
    J_initial_vector,
    tspan,
    sensitivity_parameters
)

J_sol = solve(
    sensitivity_problem,
    Tsit5();
    abstol = 1e-10,
    reltol = 1e-10,
    saveat = 0.02
)
```

The resulting `J_sol` is itself an interpolating solution. For example, `reshape(J_sol(10.0), 3, 3)` returns the complete initial-condition sensitivity matrix at $t=10$.

```julia
J_at_10 = reshape(J_sol(10.0), 3, 3)
println("Sensitivity Jacobian J(10):")
display(J_at_10)

println("Check at the initial time; this should be the identity matrix:")
display(reshape(J_sol(0.0), 3, 3))
```

## 5. Singular values of the sensitivity Jacobian

If an initial perturbation is $\delta\mathbf{x}_0$, its linearized evolution is

$$\delta\mathbf{x}(t)\approx J(t)\,\delta\mathbf{x}_0.$$

The singular values $\sigma_1(t)\geq\sigma_2(t)\geq\sigma_3(t)\geq0$ are the principal amplification factors. In particular, $\sigma_1$ is the greatest possible amplification of a unit initial perturbation. We use a logarithmic vertical axis because a chaotic trajectory can produce amplification and contraction over very different scales.

```julia
times = J_sol.t
singular_values = zeros(3, length(times))

for k in eachindex(times)
    J = reshape(J_sol.u[k], 3, 3)
    singular_values[:, k] .= svdvals(J)
end

singular_value_plot = plot(
    times, singular_values[1, :];
    label = "σ₁(t)", color = :red, linewidth = 2,
    xlabel = "time t", ylabel = "singular value",
    title = "Singular values of J(t)",
    yscale = :log10, legend = :topleft,
    grid = true, minorgrid = true, size = (850, 500)
)
plot!(times, singular_values[2, :]; label = "σ₂(t)", color = :blue, linewidth = 2)
plot!(times, singular_values[3, :]; label = "σ₃(t)", color = :green, linewidth = 2)
singular_value_plot
```

## 6. Optional: finite-time stretching rates

A related diagnostic is $\lambda_i(t)=\log(\sigma_i(t))/t$. These are finite-time stretching rates from the fixed initial time, not automatically the asymptotic Lyapunov exponents. Long-time Lyapunov exponent computations normally use repeated QR orthonormalization because directly propagated sensitivity matrices eventually become extremely ill-conditioned.

```julia
finite_time_rates = fill(NaN, 3, length(times))

for k in 2:length(times)
    finite_time_rates[:, k] .= log.(singular_values[:, k]) ./ times[k]
end

rate_plot = plot(
    times[2:end], finite_time_rates[1, 2:end];
    label = "λ₁(t)", color = :red, linewidth = 2,
    xlabel = "time t", ylabel = "log(σᵢ(t))/t",
    title = "Finite-time sensitivity growth rates",
    legend = :topright, grid = true, size = (850, 500)
)
plot!(times[2:end], finite_time_rates[2, 2:end]; label = "λ₂(t)", color = :blue, linewidth = 2)
plot!(times[2:end], finite_time_rates[3, 2:end]; label = "λ₃(t)", color = :green, linewidth = 2)
hline!([0.0]; label = "zero", color = :black, linestyle = :dash)
rate_plot
```

## 7. Important numerical interpretation

This two-solve construction is intentionally transparent, but it also exposes an important numerical point. The second calculation uses an approximation to the first trajectory, not its exact values. Consequently, the accuracy of $J(t)$ depends on both solves: errors in `state_sol(t)` perturb the coefficients in the variational equation, and the second solver introduces its own integration error. Tight first-solve tolerances and dense output help keep this controlled.

For moderate times this is an excellent way to study the tangent dynamics. For very long times, the columns of $J$ tend to align with the most unstable direction, the matrix becomes ill-conditioned, and the smallest singular values can lose relative accuracy. That is a property of the numerical formulation in a chaotic system, not necessarily a failure of `Tsit5`.
